```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)
plt.figure(figsize=(4,4))
plt.imshow(digits.images[5], cmap ='gray')
plt.title(f"Label:{digits.target[0]}")
plt.axis('off')
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
fig, axes = plt.subplots(5, 2, figsize=(10, 5))
for i, ax in enumerate (axes.flat):
    ax.imshow(digits.images[i], cmap='gray')
    ax.set_title(f"Digit:{digits.target[i]}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```

write a python program to import iris dataset from scykit_learn and convert it into pandas dataframe and display the shape of the dataset.

```python
from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(df.head())
print("Shape of the dataset:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```
